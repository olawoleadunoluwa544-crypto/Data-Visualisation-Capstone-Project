```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('heart.csv')
df.head()
```

```python
print('Shape of dataset:', df.shape)
print('Column names:', df.columns.tolist())
df.describe()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.hist(figsize=(12, 10))
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 8))
sns.boxplot(data=df)
plt.xticks()
plt.title('Box Plot for Outlier Detection')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.barplot(data=df, x='sex', y='chol', hue='target')
plt.title('Cholestorol Levels by Sex Grouped by Target')
plt.xlabel('Sex')
plt.ylabel('Cholesterol')
plt.show()
```

```python
plt.figure(figsize=(15,4))
plt.subplot(1,3,1)
sns.countplot(data=df, x='sex')
plt.title('Countplot of Sex')

plt.subplot(1,3,3)
sns.countplot(data=df, x='thal')
plt.title('Countplot of Thal')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(df.corr(), annot=True, cmap='autumn', fmt='.2f')
plt.title('Corrolation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(13,5))
plt.subplot(1,3,1)
sns.countplot(data=df, x='target')
plt.title('Distribution of Target')
plt.xlabel('Target')
plt.ylabel('Count')

plt.subplot(1,3,2)
sns.countplot(data=df, x='sex', hue='target')
plt.title('Sex vs Target Distribution')
plt.xlabel('Sex')
plt.ylabel('Count')

plt.subplot(1,3,3)
sns.countplot(data=df, x='ca', hue='target')
plt.title('CA vs Target Distribution')
plt.xlabel('CA')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```
